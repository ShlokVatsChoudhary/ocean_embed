# Day 1 - Inspect GLORYS / DUACS / OSTIA

Download a 3-day window, regrid to 0.25 deg, build X/Y, plot T at 100 m.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
from src import config, download, harmonize, viz
import numpy as np
```

## 1. Download (3-day tiny test)

```python
paths = download.download_all("2020-01-01", "2020-01-03")
paths
```

## 2. Inspect raw files

```python
import xarray as xr
for k, p in paths.items():
    ds = xr.open_dataset(p)
    print(k, dict(ds.sizes), list(ds.data_vars))
    ds.close()
```

## 3. Harmonise -> X, Y

```python
res = harmonize.build_arrays(paths["glorys"], paths["duacs"], paths["ostia"], tag="demo")
print("X", res["X"].shape, "Y", res["Y"].shape, res["channels"])
```

## 4. Plot T at 100 m

```python
zi = res["depths"].index(100)
viz.plot_field(res["Y"][0, zi], res["lat"], res["lon"],
               title="T @ 100 m", cbar_label="deg C",
               fname=config.FIGURES / "nb_T100m.png")
plt = __import__("matplotlib.pyplot", fromlist=["pyplot"]); plt.show()
```
